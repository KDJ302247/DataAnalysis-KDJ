```python
import numpy as np
import matplotlib.pyplot as plt

# 기본적인 정수 리스트 5개씩 사용
k = np.array([1, 2, 3, 4, 5])
plt.plot(k, k, 'k', label = 'one')        # k 검은색 실선
plt.plot(k, k * 2, 'r--', label = 'two')  # r-- 빨간색 파선
plt.plot(k, k * 3, 'g.', label = 'three') # g. 선 없이 초록색 점 마커로 표시
plt.legend()
plt.show()
```

코드 3-22 KDJ

```python
import matplotlib.pyplot as plt
import pandas as pd

# 데이터를 데이터프레임 형태로 읽어 오기
df = pd.read_csv('Temp_KDJ.csv', encoding = 'cp949')
df.head(3)
```

코드 3-23 KDJ

```python
df.info()
```

코드 3-24 KDJ

```python
df2 = df.fillna(method = 'ffill')
df2.info()
```

코드 3-25 KDJ

```python
df2.rename(columns = {'평균기온(°C)':'avg_temp'}, inplace = True)
df2.rename(columns = {'최고기온(°C)':'max_temp'}, inplace = True)
df2.info()
```

코드 3-26 KDJ

```python
df2['일시'] = pd.to_datetime(df2['일시'], format = '%Y-%m-%d')
df2.info()
```

코드 3-27 KDJ

```python
df3 = df2.set_index('일시')
df3.head(3)
```

코드 3-28 KDJ

```python
df_Sep = df3[pd.DatetimeIndex(df3.index).month == 9]
df_Sep
```

코드 3-29 KDJ

```python
plt.title('Seoul 2024. 9')
plt.plot(range(0, 30), df_Sep['max_temp'], label = 'max', c = 'r')
plt.plot(range(0, 30), df_Sep['avg_temp'], label = 'avg', c = 'y')

plt.xlim(1, 30)
plt.ylim(20, 37)
plt.legend()
plt.rcParams['figure.figsize'] = (8, 8)   # 그림의 가로, 세로 사이즈를 설정
plt.show()
```
